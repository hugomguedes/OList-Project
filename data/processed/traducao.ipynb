### traduzindo e salvando os nomes das tabelas do csv da pasta ./processed

```python
from pathlib import Path
import pandas as pd

raw_dir = Path("../raw")
processed_dir = Path("../processed")
processed_dir.mkdir(exist_ok=True)

traducoes = {
    "olist_sellers_dataset.csv": "vendedores.csv",
    "olist_products_dataset.csv": "produtos.csv",
    "olist_orders_dataset.csv": "pedidos.csv",
    "olist_order_reviews_dataset.csv": "avaliacoes_pedidos.csv",
    "olist_order_payments_dataset.csv": "pagamentos_pedidos.csv",
    "olist_order_items_dataset.csv": "itens_pedidos.csv",
    "olist_geolocation_dataset.csv": "geolocalizacao.csv",
    "product_category_name_translation.csv": "categorias_produtos.csv",
    "olist_customers_dataset.csv": "clientes.csv"
}

for arquivo in raw_dir.glob("*.csv"):
    df = pd.read_csv(arquivo)
    novo_nome = traducoes.get(arquivo.name, arquivo.name)
    df.to_csv(processed_dir / novo_nome, index=False)

print("CSV traduzidos salvos em ../processed!")
```

### Traduzindo os cabeçalhos de dentro da pasta ../processed

```python
import pandas as pd
from pathlib import Path

pasta = Path("../processed")

traducao = {
    # Avaliações
    "review_creation_date": "data_criacao_avaliacao",
    "review_answer_timestamp": "data_resposta_avaliacao",

    # Categorias
    "product_category_name_english": "categoria_produto_ingles",

    # Clientes
    "customer_zip_code_prefix": "cep_cliente",

    # Geolocalização
    "seller_zip_code_prefix": "cep_vendedor",
    "geolocation_zip_code_prefix": "cep",
    "geolocation_lat": "latitude",
    "geolocation_lng": "longitude",
    "geolocation_city": "cidade",
    "geolocation_state": "estado",

    # Vendedores
    "seller_zip_code_prefix": "cep_vendedor",
    "seller_city": "cidade_vendedor",
    "seller_state": "estado_vendedor",

    # Produtos
    "product_name_lenght": "tamanho_nome_produto",
    "product_description_lenght": "tamanho_descricao_produto",
    "product_photos_qty": "quantidade_fotos",
    "product_weight_g": "peso_gramas",
    "product_length_cm": "comprimento_cm",
    "product_height_cm": "altura_cm",
    "product_width_cm": "largura_cm",

    # Pedidos
    "order_id": "id_pedido",
    "customer_id": "id_cliente",
    "order_status": "status_pedido",
    "order_purchase_timestamp": "data_compra",
    "order_approved_at": "data_aprovacao",
    "order_delivered_carrier_date": "data_envio",
    "order_delivered_customer_date": "data_entrega",
    "order_estimated_delivery_date": "data_entrega_estimada",

    # Itens
    "order_item_id": "id_item_pedido",
    "product_id": "id_produto",
    "seller_id": "id_vendedor",
    "shipping_limit_date": "limite_envio",
    "price": "preco",
    "freight_value": "valor_frete",

    # Pagamentos
    "payment_sequential": "sequencia_pagamento",
    "payment_type": "tipo_pagamento",
    "payment_installments": "parcelas",
    "payment_value": "valor_pagamento"
}

for arquivo in pasta.glob("*.csv"):
    df = pd.read_csv(arquivo)

    # Traduz os cabeçalhos encontrados
    df.rename(columns=traducao, inplace=True)

    # Salva novamente o arquivo
    df.to_csv(arquivo, index=False, encoding="utf-8-sig")

    print(f"✅ Corrigido: {arquivo.name}")

print("\nTodos os cabeçalhos foram padronizados!")
```
